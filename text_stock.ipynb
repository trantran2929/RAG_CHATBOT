```python
# tests/test_stock_realtime.py
import pprint
from modules.api.stock_api import (
    get_stock_quote,
    get_history_prices,
    format_stock_info,
    format_history_text,
)

TEST_SYMBOL = "FPT"  # bạn có thể đổi sang HPG, FPT, VNM, SSI,...

def test_realtime_quote():
    print("=== [TEST] Giá hiện tại ===")
    data = get_stock_quote(TEST_SYMBOL)
    pprint.pprint(data)

    print("\n--- Format stock info ---")
    print(format_stock_info(TEST_SYMBOL))
    print("\n")

def test_history_5d():
    print("=== [TEST] Lịch sử 5 ngày ===")
    hist = get_history_prices(TEST_SYMBOL, days=5)
    pprint.pprint(hist)

    print("\n--- Format history text ---")
    print(format_history_text(TEST_SYMBOL, 5, hist))
    print("\n")

if __name__ == "__main__":
    test_realtime_quote()
    test_history_5d()
```

```python
# tests/test_intraday_and_momentum.py
import numpy as np
from modules.api.stock_api import get_intraday_df

TEST_SYMBOL = "VCB"

def test_intraday_momentum():
    print("=== [TEST] Intraday momentum ===")

    df = get_intraday_df(TEST_SYMBOL, interval="5m", days=1, debug=True)
    if df is None or df.empty:
        print("❌ Không có intraday data.")
        return

    print("OHLCV tail:")
    print(df.tail())

    close = df["close"].astype("float64").dropna()
    if len(close) < 6:
        print("Không đủ nến để tính momentum 5 bước.")
        return

    # log-return giữa các nến
    ret = np.log(close / close.shift(1)).dropna()
    last_returns = ret.tail(5)
    mu = float(last_returns.mean())
    sigma = float(last_returns.std() or 0.0)

    last_px = float(close.iloc[-1])

    print("\n--- Momentum stats ---")
    print(f"ret mean (5 nến cuối): {mu:.6f}")
    print(f"ret std  (5 nến cuối): {sigma:.6f}")
    print(f"last_px: {last_px:.3f}")

    # giả sử giữ nguyên drift mu cho 3 bước tiếp
    step_prices = []
    cur = last_px
    for i in range(3):
        cur = cur * np.exp(mu)
        step_prices.append(cur)

    print("\nDự phóng 3 bước tiếp theo (giả định momentum giữ nguyên):")
    for i,px in enumerate(step_prices, start=1):
        print(f"t+{i}: {px:.3f}")

if __name__ == "__main__":
    test_intraday_momentum()
```
